```python
# -----------------------------
# Assignment 11: NLP Preprocessing
# Using dummy Customer_Feedback column
# Works without NLTK downloads
# -----------------------------

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
import pickle

# -----------------------------
# Step 1: Load dataset
# -----------------------------
df = pd.read_csv("customer_churn_dataset.csv")

# -----------------------------
# Step 2: Create dummy text column
# -----------------------------
dummy_feedback = [
    "Very satisfied with the service",
    "Had issues with payment",
    "Support team was helpful",
    "Late delivery and support delays",
    "Loyal customer, no complaints"
]

df['Customer_Feedback'] = (dummy_feedback * ((len(df) // len(dummy_feedback)) + 1))[:len(df)]

print("Sample Customer_Feedback:")
print(df['Customer_Feedback'].head())

# -----------------------------
# Step 3: Simple preprocessing
# Lowercase, remove non-alpha words
# -----------------------------
stop_words = set([
    "the", "is", "with", "and", "a", "an", "was", "had", "very", "no"
])  # Basic stopwords list

def preprocess_text_simple(text):
    text = text.lower()
    tokens = text.split()  # Simple split on space
    tokens = [t for t in tokens if t.isalpha() and t not in stop_words]
    return ' '.join(tokens)

df['Clean_Feedback'] = df['Customer_Feedback'].apply(preprocess_text_simple)

print("\nSample Clean_Feedback:")
print(df['Clean_Feedback'].head())

# -----------------------------
# Step 4: TF-IDF Vectorization
# -----------------------------
vectorizer = TfidfVectorizer()
X_tfidf = vectorizer.fit_transform(df['Clean_Feedback'])

print("\nTF-IDF matrix shape:", X_tfidf.shape)

# -----------------------------
# Step 5: Save TF-IDF matrix and features
# -----------------------------
with open('tfidf_matrix.pkl', 'wb') as f:
    pickle.dump(X_tfidf, f)

with open('tfidf_features.pkl', 'wb') as f:
    pickle.dump(vectorizer.get_feature_names_out(), f)

print("\nTF-IDF matrix and features saved successfully!")
```
